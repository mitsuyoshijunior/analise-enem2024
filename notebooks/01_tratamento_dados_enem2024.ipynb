```python
import pandas as pd
import numpy as np
```

```python
# Lendo tabela geral do ENEM 2024
caminho = '/content/drive/MyDrive/analise-diagnostico-enem2024/RESULTADOS_2024.csv'
df = pd.read_csv(caminho, sep = ';', encoding='latin-1')
```

## Tratando dados

```python
# Identificando colunas
df.columns
```

```python
# Atribuindo as colunas desejadas
df = df[['NU_SEQUENCIAL','NO_MUNICIPIO_PROVA', 'SG_UF_PROVA', 'TP_DEPENDENCIA_ADM_ESC', 'TP_LOCALIZACAO_ESC', 'NU_NOTA_CN', 'NU_NOTA_CH', 'NU_NOTA_LC', 'NU_NOTA_MT', 'NU_NOTA_REDACAO']]
```

```python
# Verificando se tem duplicatas
df.duplicated().sum()
```

```python
# Excluindo linhas com notas faltantes
df = df.dropna(subset=['NU_NOTA_CN', 'NU_NOTA_CH', 'NU_NOTA_LC', 'NU_NOTA_MT', 'NU_NOTA_REDACAO'])
```

```python
# Resetando o índice
df = df.reset_index(drop=True)
```

```python
# Somando dados faltantes
df.isna().sum()
```

```python
# Atualizando nome das colunas por nomes mais intuitivos
df = df.rename(columns={
  'NU_SEQUENCIAL':'numero_seq_aluno',
  'NO_MUNICIPIO_PROVA':'municipio',
  'SG_UF_PROVA':'uf',
  'TP_DEPENDENCIA_ADM_ESC':'dependencia',
  'TP_LOCALIZACAO_ESC':'localizacao',
  'NU_NOTA_CN':'nota_ciencias_naturais',
  'NU_NOTA_CH':'nota_ciencias_humanas',
  'NU_NOTA_LC':'nota_linguagens',
  'NU_NOTA_MT':'nota_matematica',
  'NU_NOTA_REDACAO':'nota_redacao'
  })
```

```python
# Verificando o tipo de dados das colunas para possível tratamento
df.dtypes
```

```python
# Tratando dados objects nulos
df['dependencia'] = df['dependencia'].fillna('Não informado')
df['localizacao'] = df['localizacao'].fillna('Não informado')
```

```python
# Substituindo valores da coluna 'dependencia'
df['dependencia'] = df['dependencia'].replace({
    1:'Federal',
    2:'Estadual',
    3:'Municipal',
    4:'Privada'
})
```

```python
# Substituindo valores da coluna 'localizacao'
df['localizacao'] = df['localizacao'].replace({
    1:'Urbana',
    2:'Rural'
})
```

```python
# Verificação final
print(df.info())
print(df.describe(include='all'))
print('Duplicatas restantes:', df.duplicated().sum())
```

```python
# Criando caminho de saída
caminho_saida_tabela = '/content/drive/MyDrive/analise-diagnostico-enem2024/tabela_enem2024.csv'
```

```python
df.to_csv(caminho_saida_tabela, index=False)
```
