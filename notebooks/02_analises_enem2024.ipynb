## Calculando média e desvio padrão por estado

```python
# Calculando média por estado
df_media_uf = df.groupby('uf')['media_geral'].mean().round(2).reset_index()
```

```python
# Calculando o desvio padrão e adicionando coluna a tabela
desvio_por_uf = df.groupby('uf')['media_geral'].std().round(2)
df_media_uf['desvio_padrao'] = df_media_uf['uf'].map(desvio_por_uf).round(2)
```

## Calculando média e desvio padrão por região

```python
# Definindo listas de regiões
norte = ['AC', 'AP', 'AM', 'PA', 'RO', 'RR', 'TO']
nordeste = ['AL', 'BA', 'CE', 'MA', 'PB', 'PE', 'PI', 'RN', 'SE']
centro_oeste = ['DF', 'GO', 'MT', 'MS']
sudeste = ['ES', 'MG', 'RJ', 'SP']
sul = ['PR', 'RS', 'SC']
```

```python
# Criando função para identificar a região em que a UF está
regioes = {
    'N': norte,
    'NE': nordeste,
    'CO': centro_oeste,
    'SE': sudeste,
    'S': sul
}

def definir_regiao(uf):
    for regiao, lista_ufs in regioes.items():
        if uf in lista_ufs:
            return regiao
    return 'Desconhecida'  # caso o UF não esteja em nenhuma lista
```

```python
# Criando a coluna de identificação da região
df_media_uf['regiao'] = df_media_uf['uf'].apply(definir_regiao)
```

```python
# Exibindo pequena amostra
df_media_uf.head()
```

```python
# Criando df de média por região
df_media_regiao = df_media_uf.groupby('regiao')['media_geral'].mean().round(2)
```

```python
# Ordenando da maior pra menor média
df_media_regiao = df_media_regiao.sort_values(ascending=False).reset_index()
```

```python
# Exibindo média por região
df_media_regiao
```

```python
# Criando df de desvio padrão por região
df_desvio_por_regiao = df_media_uf.groupby('regiao')['media_geral'].std().round(2).rename('desvio_padrao')
```

```python
# Exibindo tabela
df_desvio_por_regiao
```

```python
# Fazendo a junção das tabelas

df_media_e_desvio_por_regiao = df_media_regiao.merge(df_desvio_por_regiao,
                  how = 'inner',
                  left_on = ['regiao'],
                  right_on = ['regiao'])
```

```python
# Exibindo nova df de média e desvio por região
df_media_e_desvio_por_regiao
```

```python
# Checando tipagem de daos
df_media_e_desvio_por_regiao.dtypes
```

```python
# Salvando tabela
df_media_e_desvio_por_regiao.to_csv('/content/drive/MyDrive/analise-diagnostico-enem2024/consultas/df_media_desvio_por_regiao.csv', decimal=',', index=False)
```

## Calculando porcentagem de alunos com acima de 600 pontos de média por estado

```python
# Definindo função de triagem
media = 600
def acima_media(x):
  if x > media:
    return True
  else:
    return False
```

```python
# Copiando a tabela principal
df2 = df.copy()
```

```python
# Criando coluna de identificação de limite da nota
df2['media_maior_que_600'] = df2['media_geral'].apply(acima_media)
```

```python
# Exibindo amostra da nova tabela
df2.head()
```

```python
# Criando nova series com porcentagem de nota média acima de 600 pontos
porcentagem_media_600_por_estado = (df2.groupby('uf')['media_maior_que_600'].mean()*100).rename('porcentagem_acima_de_600')
```

```python
# Arredondando porcentagem para até duas casas decimais e ordenando do maior para o menor
porcentagem_media_600_por_estado = porcentagem_media_600_por_estado.round(2).sort_values(ascending=False).reset_index()

# Ajustando indice para começar em 1
porcentagem_media_600_por_estado.index += 1
```

```python
# Exibindo tabela
porcentagem_media_600_por_estado
```

```python
# Checando tipagem de dados
porcentagem_media_600_por_estado.dtypes
```

```python
# Salvando df
porcentagem_media_600_por_estado.to_csv('/content/drive/MyDrive/analise-diagnostico-enem2024/consultas/porcentagem_media600_por_estado.csv', decimal=',', index=False)
```

## Calculando quantidade de alunos com mais de 600 pontos
